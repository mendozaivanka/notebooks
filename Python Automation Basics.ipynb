````python
# Built-in function:
print("Hello World")

# User-defined function:
def greet(name):
    print("Hello,", name)

greet("Audience")
````

### Packages
A package is a collection of modules, functions, and tools that provide additional functionality beyond Python's built-in features.

* **math:** Mathematical computations
* **pandas:** Data analysis
* **numpy:** Numerical Computation
* **matplotlib:** Data Visualization

````python
# EXAMPLES
# How to import and use a built-in package
import math
import datetime

# Using the math package to calculate a square root
root = math.sqrt(64)
print(f"The square root of 64 is: {root}")

# Using the datetime package to get today's date
today = datetime.date.today()
print(f"Today's date is: {today}")
````

### Exceptions
An exception is an error that occurs during program execution, interrupting the normal flow of the program. It helps developers identify and handle failure gracefully instead of allowing
the program to crash.

````python
# try:
    # Code that may cause an error
# except:
    # Code that handles the error
````

````python
# Demonstrating how to handle an exception without crashing the program
try:
    # Trying to divide by zero (this will cause an error)
    result = 10 / 5
except ZeroDivisionError:
    print("Oops! You cannot divide by zero. Continuing script...")
    
print("The script is still running!")
````

### Environment Management
It is the practice of storing and managing application settings, credentials, and configuration values outside of the source code.
Instead of hardcoding values in Python scripts, these values are stored as environment variables and loaded when the application runs.

**Common Configuration Values**
* API Keys
* Database Passwords
* Database Hosts
* Connection Strings
* Application URLs
* Cloud Credentials

### **.env file:** 
* A text file used to store environment variables and configuration settings for an application.Typically, it is placed in the project's root folder and is not committed to source control.
### **dotenv:** 
* A Python package that loads environment variables from a .env file into your application. It helps keep sensitive information (such as passwords, API keys, and connection strings) separate from your code.
### **OS:** 
* A built-in Python package that allows your program to interact with the operating system. It can be used to access environment variables, work with files and folders, get system information, and manage file paths.

````python
# Example of a .env file
# API_KEY=123456
# DB_HOST=localhost
# DB_USER=admin
# DB_PASSWORD=password123
````

## 5. What is ETL? (Extract, Transform, Load)
* **Extract:** Fetching raw data from various sources.
* **Transform:** Processing of raw data which includes but is not limited to cleaning, deduplicating, aggregating, and performing calculations.
* **Load:** Storing the processed data into its final destination.
